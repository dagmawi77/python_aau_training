```python
meters = pd.read_csv('aaus1_class_meter_readings - aaus1_class_meter_readings.csv',parse_dates=['reading_date'])
print(meters.shape)
print(meters['branch'].nunique())
print(meters.usage_m3.isna().sum())
meters.head()
meters.dtypes
meters.info()
meters.tail
print(meters.columns[:])
sum_meter=sum(meters['current_m3'])
count_record_id = len(meters['record_id'])
print("Number Of Record : ",count_record_id)
print("The Sum Of the Total meters is : ",sum_meter)
print(meters.size)
print(meters.index.start,meters.index.stop)

missing =  meters.isna().sum()
print(missing[missing>0].to_string())
print("Complete Rows : ", len(meters.dropna()))

```

```python
b = np.linspace(0,3,1000)
print(b)
```

```python
usage = meters.usage_m3.to_numpy()
valid = usage[~np.isnan(usage)]
print(round(np.mean(valid),2))
print(np.median(valid))
print(round(np.std(valid),2))
```

```python
monthly = meters.groupby(
  'reading_date'
)['usage_m3'].sum()
print(monthly.astype(int).tolist())
print(monthly.index.month.tolist())
```

```python
import matplotlib.pyplot as plt
monthly.plot.line(marker='o')
plt.title('Monthly consumption')
plt.xlabel('Reading month')
plt.ylabel('Recorded volume (m³)')
plt.tight_layout()
plt.show()
```

```python
cases = pd.read_csv(
'aaus1_class_service_requests.csv')
print(cases.shape)
print(cases.branch.value_counts()
.sort_index().to_string())
```

```python
counts = cases.branch.value_counts()
counts = counts.sort_values()
counts.plot.barh()
plt.xlabel('Number of requests')
plt.ylabel('Branch')
plt.tight_layout()
plt.show()
```

```python
counts = cases.branch.value_counts()
counts = counts.sort_values()
counts.plot.barh()
plt.xlabel('Number of requests')
plt.ylabel('Branch')
plt.tight_layout()
plt.show()
```
