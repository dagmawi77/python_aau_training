```python
bill= 1000
tp = 0
for i in range(10):
    cp = int(input("Enter the amount paid by customer: "))
    tp += cp
    if (tp <= bill*0.5):
        print(f"Customer has paid {tp} more than 50% of the bill.")
    elif (tp <= bill)*.75:
        print(f"Customer has paid {tp} more than 75% of the bill.")
    elif (tp <= bill)*.9:
        print(f"Customer has paid {tp} more than 90% of the bill.")
    elif (tp == bill):
        print(f"Customer has paid the full bill amount of {bill}.")
        break
    else:
        print(f"Customer has paid {tp} which is less than 50% of the bill.")
    
```
