# Water Bill Calculation

```python
prrivues_reading = int(input("Please enter the previous reading: "))
current_reading = int(input("Please enter the current reading: "))
consmption = current_reading - prrivues_reading
print(f"The consumption is {consmption} m3.")
customer_type = input("Please enter the customer type (dom,non-dom,fountain): ")
meter_type = input("Please enter the meter type (0.5,0.75.1,2,3,4,5,6): ")
tariff_block1= 5
tarriff_block2= 10
tariff_block3= 15
tarriff_block4= 20
if customer_type == "dom":
    if consmption <= 10:
        water_cons_amount = consmption * tariff_block1
        print(f"The total payment is {water_cons_amount}. and a customer block is block 1.")
    elif consmption <= 20:
        water_cons_amount = (10 * tariff_block1) + ((consmption - 10) * tarriff_block2)
        print(f"The total payment is {water_cons_amount}. and a customer block is block 2.")
    elif consmption <= 30:
        water_cons_amount = (10 * tariff_block1) + (10 * tarriff_block2) + ((consmption - 20) * tariff_block3)
        print(f"The total payment is {water_cons_amount}. and a customer block is block 3.")
    else:
        water_cons_amount = (10 * tariff_block1) + (10 * tarriff_block2) + (10 * tariff_block3) + ((consmption - 30) * tarriff_block4)
        print(f"The total payment is {water_cons_amount}. and a customer block is block 4.")
sanitaion_fee = water_cons_amount * 0.7
mentance_fee = water_cons_amount * 0.1
disater_fee = 5

if meter_type == 0.5:
    meter_fee = 10
elif meter_type == 0.75:
    meter_fee = 15
elif meter_type == 1:
    meter_fee = 20
elif meter_type == 2:
    meter_fee = 30
elif meter_type == 3:
    meter_fee = 40
elif meter_type == 4:
    meter_fee = 50
elif meter_type == 5:
    meter_fee = 60
elif meter_type == 6:
    meter_fee = 70
else:
    meter_fee = 0

grand_total = water_cons_amount + sanitaion_fee + mentance_fee + disater_fee + meter_fee
print(f"The water consumption amount is {water_cons_amount}.")
print(f"The sanitaion fee is {sanitaion_fee}.")
print(f"The mentance fee is {mentance_fee}.")
print(f"The disater fee is {disater_fee}.")
print(f"The meter fee is {meter_fee}.")
print(f"The grand total is {grand_total}.")

```
