# Type Conversion

```python
age_text = "25"
age = int(age_text)
print(f"The age is {age} and the type of age is {type(age)}")
name = int(3.9)
print(f"The name is {name} and the type of name is {type(name)}")
boolean_value = bool("")
print(f"The boolean value is {boolean_value} and the type of boolean value is {type(boolean_value)}")
namefloat = int(3.14)
print(f"The namefloat is {namefloat} and the type of namefloat is {type(namefloat)}")
```
