```python
import csv

def csvReader(filename):
    lines = []
    with open(filename, "r", encoding="utf-8") as f:
        reader = csv.DictReader(f)
        for row in reader:
            print(row)

file = csv.reader(open("bill.csv", "r", encoding="utf-8"))
for row in file:
    print(row)

def calcBill():
    custdata = csv.DictReader(file)
    for d in custdata:
        if int(d["CURRREAD"]-d["PREVREAD"])<100:
            d["BILL"] = int(d["CONS"])*120
        elif int(d["CURRREAD"]-d["PREVREAD"])>=100:
            d["BILL"] = int(d["CONS"])*150
        else:
            d["BILL"] = 0
calcBill()
csv.writer(open("bill.csv", "a", encoding="utf-8")).writerows(file)
```
