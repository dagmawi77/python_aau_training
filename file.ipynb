```python
file = open("file.txt", "r",encoding="utf-8")
content = file.read()
print(content)

file.close()
filewrite = open("abcd.txt", "a",encoding="utf-8")
filewrite.write("Hello, World! 123")
filewrite.close()

with open("abcd.txt", "r",encoding="utf-8") as file:
    content = file.read()
    print(content)

def fileWriter(filename, text):
    with open(filename, "a", encoding="utf-8") as file:
        file.write(text)


input("Press Enter to continue...")
input_text = input("Enter text to write to the file: ")
fileWriter("abcd.txt", input_text)
input("Press Enter to continue...")
input_filename = input("Enter the filename to read from: ")
fileWriter(input_filename, input_text)

```
