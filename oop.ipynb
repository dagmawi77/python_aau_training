```python
class PartyAnimal:
    nameOfAnimal = ""
    lifeSpanOfAnimal = 0
    ColorofAnimal = []
    def __init__(self, name, lifespan, color):
        self.nameOfAnimal = name
        self.lifeSpanOfAnimal = lifespan
        self.ColorofAnimal = color
    def chracteristics(self):
        return f"The name of the animal is {self.nameOfAnimal}, its lifespan is {self.lifeSpanOfAnimal} years, and its colors are {', '.join(self.ColorofAnimal)}."

lion = PartyAnimal("Lion", 15, ["Golden", "Brown"])
print(lion.chracteristics())
cat = PartyAnimal("Cat", 20, ["Black", "White"])
print(cat.chracteristics())
```
