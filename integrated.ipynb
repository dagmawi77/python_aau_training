```python
# DATA_CREATOR: SESSION9
import csv

categories = ["Leak", "Billing", "Meter", "Connection"]
branches = ["North", "Central", "West", "East"]
status_cycle = ["Open", "Closed", "Assigned", " open ", "OPEN", "Closed"]

synthetic_rows = [
    ["request_id", "category", "branch", "status", "priority"]
]

for i in range(1, 37):
    request_id = f"SR-{9000 + i}"
    category = categories[(i - 1) % len(categories)]
    branch = branches[(i - 1) % len(branches)]
    status = status_cycle[(i - 1) % len(status_cycle)]
    if i % 9 == 0:
        priority = "missing"
    else:
        priority = str((i % 5) + 1)
    synthetic_rows.append([request_id, category, branch, status, priority])

with open("session9_complaints.csv", "w", newline="", encoding="utf-8") as f:
    writer = csv.writer(f)
    writer.writerows(synthetic_rows)

synthetic_record_count = len(synthetic_rows) - 1
print("Created session9_complaints.csv")
print("Synthetic records:", synthetic_record_count)
```
