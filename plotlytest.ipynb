```python
import pandas as pd
import plotly.graph_objects as go
import plotly.express as px
import plotly.subplots as sp
```

```python
s = pd.read_csv('vizs3_service_weekly.csv')
w = pd.read_csv('vizs3_waterpoints.csv')
t = pd.read_csv('vizs3_tanker_dispatch.csv',parse_dates=['dispatch_date'])
p = pd.read_csv('vizs3_capital_projects.csv')
print(len(s), len(w), len(t), len(p))
print(s.shape)
print(w.shape)
print(t.shape)
print(p.shape)
print(s['zone'].unique())
print(t['priority'].unique().tolist())
print(w['status'].unique().tolist())
```

```python
s['week_start'] = pd.to_datetime(
s['week_start'])
weekly = s.groupby('week_start')[
'reported_cases'].sum().reset_index()
fig = px.line(weekly, x='week_start',
y='reported_cases', markers=True)
fig.show()
```

```python
z = s.groupby('zone', as_index=False).agg(
avg_hours=('avg_response_hours', 'mean'),
cases=('reported_cases', 'sum'))
fig = px.bar(z.sort_values('avg_hours'),
x='zone', y='avg_hours', color='cases')
fig.show()
```

```python
z = s.groupby('zone', as_index=False).agg(
avg_hours=('avg_response_hours', 'mean'),
cases=('reported_cases', 'sum'))
fig = px.bar(z.sort_values('avg_hours'),
x='zone', y='avg_hours', color=['green', 'orange', 'red', 'blue','purple','pink'])
fig.show()
```

```python
z = s.groupby('zone', as_index=False).agg(
avg_hours=('avg_response_hours', 'mean'),
cases=('reported_cases', 'sum'))
fig = px.bar(z.sort_values('avg_hours'),
x='zone', y='avg_hours', color=['green', 'orange', 'red', 'blue','purple','pink'])
fig.show()
```

```python
daily = s.groupby('week_start', 
as_index=False)['reported_cases'].sum()
fig = go.Figure()
fig.add_trace(go.Scatter(x=daily.week_start,
   y=daily.reported_cases, mode='lines+markers',
   name='Service cases'))
fig.update_layout(title='Weekly service cases')
```

```python
make_subplots = sp.make_subplots
fig = make_subplots(rows=1, cols=2)
fig.add_trace(go.Scatter(
x=daily.week_start, y=daily.reported_cases),
row=1, col=1)
fig.add_trace(go.Box(
x=t.priority, y=t.response_hours),
row=1, col=2)
```

```python

monthly = (t.set_index('dispatch_date').resample('MS')['delivered_liters'].sum().reset_index())
fig = px.line(monthly, x='dispatch_date',
y='delivered_liters')
fig.update_xaxes(rangeslider_visible=True)
```

```python
fig.update_layout(updatemenus=[dict(
buttons=[dict(label='All', method='relayout',
args=[{'showlegend': True}]),
dict(label='Hide legend', method='relayout',
args=[{'showlegend': False}])],
x=.02, y=1.16)])
```

```python
fig.update_traces(hovertemplate=(
'Date: %{x}<br>'
'Delivered: %{y:,.0f} L'
'<extra></extra>'))
fig.show()
```
