```python

import matplotlib.pyplot as plt

# Sample data
labels = ['Water Supply', 'Sewerage', 'Maintenance', 'Administration']
sizes = [40, 25, 20, 15]

# Create a figure with a 3D effect
fig, ax = plt.subplots(figsize=(9, 7))

# Create a 2D pie chart
wedges, texts, autotexts = ax.pie(
    sizes,
    labels=labels,
    autopct='%1.1f%%',
    startangle=90,
    explode=(0.05, 0.05, 0, 0),
    shadow=True
)

ax.set_title('AAWSA Department Budget Distribution',
             fontsize=15)

plt.tight_layout()
plt.show()
```

```python

import seaborn as sns
import matplotlib.pyplot as plt

# Apply Seaborn style
sns.set_theme(style="whitegrid", palette="pastel")

# Sample data
departments = [
    "Water Supply",
    "Sewerage",
    "Maintenance",
    "Administration"
]
values = [40, 25, 20, 15]

# Create pie chart
plt.figure(figsize=(8, 8))

plt.pie(
    values,
    labels=departments,
    autopct="%1.1f%%",
    startangle=90,
    shadow=True,
    explode=(0.05, 0, 0, 0)
)

plt.title("Department Distribution", fontsize=16)
plt.show()
```

```python
import plotly.express as px
import pandas as pd

# Sample data
df = pd.DataFrame({
    "Department": [
        "Water Supply",
        "Sewerage",
        "Maintenance",
        "Administration"
    ],
    "Budget": [400000, 250000, 200000, 150000]
})

# Create interactive pie chart
fig = px.pie(
    df,
    names="Department",
    values="Budget",
    title="AAWSA Department Budget Distribution",
    hole=0.35,  # Donut effect
    color_discrete_sequence=px.colors.qualitative.Set2,
    hover_data=["Budget"]
)

fig.update_traces(
    textposition="inside",
    textinfo="label+percent",
    pull=[0.08, 0, 0, 0],
    hovertemplate=(
        "<b>%{label}</b><br>"
        "Budget: %{value:,.0f}<br>"
        "Share: %{percent}"
        "<extra></extra>"
    )
)

fig.update_layout(
    title_x=0.5,
    template="plotly_white"
)

fig.show()
```
