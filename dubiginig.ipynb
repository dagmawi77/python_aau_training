```python
def calculate_bill(usage):
    if not isinstance(usage, (int, float)):
        raise TypeError("Usage must be a number")

    if usage < 0:
        raise ValueError("Usage cannot be negative")

    return usage * 0.70
calculate_bill(-100)  # This will raise a ValueError
calculate_bill("100")  # This will raise a TypeError
```

```python
def calculate_bill(usage, rate=0.70, service_fee=5):
    return usage * rate + service_fee


def test_calculate_bill():
    assert calculate_bill(100) == 75


def test_zero_usage():
    assert calculate_bill(0) == 5


def test_large_usage():
    assert calculate_bill(1000) == 705
```

```python
import csv

with open("Customer.csv", newline="") as file:
    reader = csv.DictReader(file)

    for row in reader:
        print("DEBUG:", row)

        customer_id = row["customer_id"]
        usage = float(row["usage"])

        print(
            "Customer:",
            customer_id,
            "Usage:",
            usage
        )
```
