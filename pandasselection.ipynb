```python
active = orders.loc[
~(orders.status == 'Closed')]
print(len(active))
print(active.status.unique().tolist())
```

```python
asset_subset = orders.loc[
orders.asset_type.isin(
['Pipe', 'Pump'])]
print(len(asset_subset))
print(sorted(asset_subset.asset_type.unique()))
```

```python
mid_cost = orders.loc[
orders.estimated_cost_etb
.between(8000, 12000)]
print(len(mid_cost))
print(mid_cost.estimated_cost_etb
.between(8000, 12000).all())
```

```python
repair = orders.loc[
orders.work_type.str.contains(
'Repair', case=False, na=False)]
print(len(repair))
print(repair.work_type.unique())
```
